```python
#ISIC 2019 Ablation C3
#imports
import os
import glob
import random
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

import cv2
import numpy as np
import pandas as pd
from collections import Counter

from PIL import Image
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader, WeightedRandomSampler
from torch.amp import autocast, GradScaler

from torchinfo import summary

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score
)

import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
# ── Reproducibility ─────────────────────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark     = False
os.environ["PYTHONHASHSEED"] = str(SEED)
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 32

train_dir = "/kaggle/input/datasets/sujansarkar/isic-2019-multi-modal-classification/Train_Validation_dataset/content/Train_Validation_dataset/train"
val_dir   = "/kaggle/input/datasets/sujansarkar/isic-2019-multi-modal-classification/Train_Validation_dataset/content/Train_Validation_dataset/val"
test_dir  = "/kaggle/input/datasets/sujansarkar/isic-2019-multi-modal-classification/Test_dataset/content/test_dataset2"

train_tfms = transforms.Compose([
    transforms.Resize(288),
    transforms.RandomResizedCrop(256, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomApply([transforms.ColorJitter(brightness=0.2, contrast=0.2, hue=0.05)], p=0.5),
    transforms.RandomAffine(degrees=180, translate=(0.1, 0.1), scale=(0.9, 1.1), shear=15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

val_test_tfms = transforms.Compose([
    transforms.Resize(288),
    transforms.CenterCrop(256),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
# Training metadata: image, age_approx, anatom_site_general, lesion_id, sex
# Test metadata:     image, age_approx, anatom_site_general, sex  (no lesion_id, no labels)
TRAIN_META_PATH = "/kaggle/input/datasets/sujansarkar/isic-2019-multi-modal-classification/ISIC_2019_Training_Metadata.csv"
TEST_META_PATH  = "/kaggle/input/datasets/sujansarkar/isic-2019-multi-modal-classification/ISIC_2019_Test_Metadata.csv"

NUMERICAL_COLS   = ['age_approx']
CATEGORICAL_COLS = ['sex', 'anatom_site_general']
PROMPT_COLS      = ['age_approx', 'sex', 'anatom_site_general']

train_meta = pd.read_csv(TRAIN_META_PATH).drop(columns=['lesion_id'], errors='ignore')
test_meta  = pd.read_csv(TEST_META_PATH)

# Compute normalization stats from training images only — no val/test leakage
train_stats = {}
for col in NUMERICAL_COLS:
    numeric = pd.to_numeric(train_meta[col], errors='coerce')
    train_stats[col] = {'mean': numeric.mean(), 'std': numeric.std() + 1e-8}

def preprocess_meta(df, train_stats):
    df = df.copy()
    df['image'] = df['image'].astype(str)
    for col in NUMERICAL_COLS:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(train_stats[col]['mean'])
        df[f"{col}_norm"] = (df[col] - train_stats[col]['mean']) / train_stats[col]['std']
    for col in CATEGORICAL_COLS:
        df[col] = df[col].fillna('unknown').astype(str)
    for col in PROMPT_COLS:
        if col in df.columns:
            df[f'raw_{col}'] = df[col].astype(str)
    return df

train_meta_p = preprocess_meta(train_meta, train_stats)
test_meta_p  = preprocess_meta(test_meta,  train_stats)

# One-hot encode — fit on training, align test to same columns
train_enc = pd.get_dummies(train_meta_p, columns=CATEGORICAL_COLS, dummy_na=False)
test_enc  = pd.get_dummies(test_meta_p,  columns=CATEGORICAL_COLS, dummy_na=False)
train_enc, test_enc = train_enc.align(test_enc, join='left', axis=1, fill_value=0)

train_enc = train_enc.set_index('image')
test_enc  = test_enc.set_index('image')

combined_meta = pd.concat([train_enc, test_enc])

META_COLS = [c for c in train_enc.columns
             if c.startswith(tuple(CATEGORICAL_COLS)) or c.endswith('_norm')]
NUM_META_FEATURES = len(META_COLS)

# Build O(1) lookup dict: image_id → pd.Series
meta_lookup = {img_id: row for img_id, row in combined_meta.iterrows()}

print(f"Tabular vector size: {NUM_META_FEATURES}")
print(f"Raw prompt cols: {[c for c in train_enc.columns if c.startswith('raw_')]}")
print(f"Metadata coverage: {len(meta_lookup)} image IDs")
```

```python
class ImageMetaDataset(datasets.ImageFolder):
    """Image + tabular metadata only — no CLIP tokenizer. C3 ablation config."""
    def __init__(self, root, meta_lookup, meta_cols, transform=None):
        super().__init__(root, transform=transform)
        self.meta_lookup = meta_lookup
        self.meta_cols   = meta_cols
        self.zero_meta   = np.zeros(len(meta_cols), dtype=np.float32)

    def __getitem__(self, index):
        image, label = super().__getitem__(index)
        path, _      = self.samples[index]
        img_name     = os.path.splitext(os.path.basename(path))[0]

        if img_name in self.meta_lookup:
            tab_vec = self.meta_lookup[img_name][self.meta_cols].values.astype(np.float32)
        else:
            tab_vec = self.zero_meta

        return image, torch.tensor(tab_vec), label
```

```python
train_ds = ImageMetaDataset(train_dir, meta_lookup, META_COLS, transform=train_tfms)
val_ds   = ImageMetaDataset(val_dir,   meta_lookup, META_COLS, transform=val_test_tfms)
test_ds  = ImageMetaDataset(test_dir,  meta_lookup, META_COLS, transform=val_test_tfms)

train_targets   = train_ds.targets
class_counts    = Counter(train_targets)
NUM_CLS         = len(class_counts)
total_samples   = sum(class_counts.values())
smoothed_counts = {cls: np.power(count, 0.5) for cls, count in class_counts.items()}
class_weights_s = {cls: 1.0 / count for cls, count in smoothed_counts.items()}
sample_weights  = torch.DoubleTensor([class_weights_s[t] for t in train_targets])

# Inverse-sqrt class weights for FocalLoss
raw_weights          = [total_samples / (NUM_CLS * class_counts[i] ** 0.5) for i in range(NUM_CLS)]
mean_w               = sum(raw_weights) / NUM_CLS
class_weights_tensor = torch.FloatTensor([w / mean_w for w in raw_weights])

def seed_worker(worker_id):
    worker_seed = SEED + worker_id
    random.seed(worker_seed)
    np.random.seed(worker_seed)
    torch.manual_seed(worker_seed)

g_sampler = torch.Generator()
g_sampler.manual_seed(SEED)
sampler = WeightedRandomSampler(
    weights=sample_weights, num_samples=len(sample_weights),
    replacement=True, generator=g_sampler
)

g_train = torch.Generator().manual_seed(SEED)
train_loader = DataLoader(
    train_ds, batch_size=BATCH_SIZE, sampler=sampler,
    num_workers=4, pin_memory=True, drop_last=True,
    worker_init_fn=seed_worker, generator=g_train
)
val_loader = DataLoader(
    val_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=4, pin_memory=True, worker_init_fn=seed_worker
)
test_loader = DataLoader(
    test_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=4, pin_memory=True, worker_init_fn=seed_worker
)
```

```python
class_to_idx = train_ds.class_to_idx

IDX_TO_CLASS = {
    class_to_idx["AK"]:   "Actinic Keratoses (AK)",
    class_to_idx["BCC"]:  "Basal Cell Carcinoma (BCC)",
    class_to_idx["BKL"]:  "Benign Keratosis-like Lesions (BKL)",
    class_to_idx["DF"]:   "Dermatofibroma (DF)",
    class_to_idx["MEL"]:  "Melanoma (MEL)",
    class_to_idx["NV"]:   "Melanocytic Nevi (NV)",
    class_to_idx["SCC"]:  "Squamous Cell Carcinoma (SCC)",
    class_to_idx["VASC"]: "Vascular Lesions (VASC)",
}

NUM_CLASSES = len(IDX_TO_CLASS)
class_names = [IDX_TO_CLASS[i] for i in range(NUM_CLASSES)]

print("Train samples:", len(train_ds))
print("Validation samples:", len(val_ds))
print("Test samples:", len(test_ds))

print("\nClasses found by ImageFolder:")
print(train_ds.classes)

print("\nClass to index mapping:")
print(train_ds.class_to_idx)

print("\nReadable class names:")
for i, name in enumerate(class_names):
    print(i, "→", name)
```

```python
class FusionNet(nn.Module):
    """
    ISIC 2019 Ablation C3 -- Image + Metadata.
    """
    def __init__(self, num_meta_features, num_classes=8, pretrained=True, freeze_backbone=False):
        super().__init__()

        # Vision: ConvNeXt-Tiny only
        convnext = models.convnext_tiny(
            weights=models.ConvNeXt_Tiny_Weights.DEFAULT if pretrained else None
        )
        self.features  = convnext.features
        self.pool      = nn.AdaptiveAvgPool2d(1)
        self.feat_dim  = 768

        # Tabular MLP
        self.tabular_mlp = nn.Sequential(
            nn.Linear(num_meta_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3)
        )
        self.meta_proj = nn.Linear(128, 256)
        self.tab_dim   = 256

        # Classifier: concat [B, 768+256=1024] -> num_classes
        self.classifier = nn.Sequential(
            nn.Dropout(0.4),
            nn.Linear(self.feat_dim + self.tab_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

        if freeze_backbone:
            for p in self.features.parameters():
                p.requires_grad = False

    def forward(self, x, meta_data):
        f_vis = self.pool(self.features(x)).flatten(1)  # [B, 768]
        f_tab = self.meta_proj(self.tabular_mlp(meta_data))  # [B, 256]
        return self.classifier(torch.cat([f_vis, f_tab], dim=1))


model = FusionNet(
    num_meta_features=NUM_META_FEATURES,
    num_classes=NUM_CLASSES,
    pretrained=True,
    freeze_backbone=True
).to(DEVICE)

summary(
    model,
    input_size=[(BATCH_SIZE, 3, 256, 256), (BATCH_SIZE, NUM_META_FEATURES)],
    dtypes=[torch.float32, torch.float32],
    col_names=["input_size", "output_size", "num_params"],
    depth=3
)
```

```python
class FocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=2.0):
        super().__init__()
        self.gamma = gamma
        self.ce = nn.CrossEntropyLoss(weight=weight, reduction="none", label_smoothing=0.1)

    def forward(self, inputs, targets):
        ce_loss    = self.ce(inputs, targets)
        pt         = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean()
```

```python
criterion = FocalLoss(weight=class_weights_tensor.to(DEVICE), gamma=2.0)

optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4, weight_decay=1e-2
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
    optimizer, T_0=10, T_mult=2, eta_min=1e-6
)
```

```python
scaler = GradScaler("cuda")

def compute_specificity(y_true, y_pred, num_classes):
    cm = confusion_matrix(y_true, y_pred, labels=range(num_classes))
    specificities = []
    for i in range(num_classes):
        TP = cm[i, i]; FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP; TN = cm.sum() - (TP + FP + FN)
        specificities.append(TN / (TN + FP) if (TN + FP) > 0 else 0)
    return np.mean(specificities)

def mixup_batch(images, meta_data, labels, alpha=0.3):
    lam      = np.random.beta(alpha, alpha)
    rand_idx = torch.randperm(images.size(0), device=images.device)
    return (lam * images + (1 - lam) * images[rand_idx],
            lam * meta_data + (1 - lam) * meta_data[rand_idx],
            labels, labels[rand_idx], lam)

def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    y_true, y_pred = [], []

    for images, meta_data, labels in tqdm(loader, desc="Training"):
        images    = images.to(device)
        meta_data = meta_data.to(device)
        labels    = labels.to(device)

        mixed_images, mixed_meta, y_a, y_b, lam = mixup_batch(images, meta_data, labels)

        optimizer.zero_grad()
        with autocast("cuda"):
            outputs = model(mixed_images, mixed_meta)
            loss    = lam * criterion(outputs, y_a) + (1 - lam) * criterion(outputs, y_b)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item() * images.size(0)
        preds = torch.argmax(outputs, 1)
        y_true.extend(y_a.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())

    avg_loss = running_loss / len(y_true)
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="macro", zero_division=0)
    rec  = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1   = f1_score(y_true, y_pred, average="macro", zero_division=0)
    spec = compute_specificity(y_true, y_pred, NUM_CLASSES)
    return avg_loss, acc, prec, rec, f1, spec

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    y_true, y_pred = [], []

    with torch.no_grad():
        for images, meta_data, labels in tqdm(loader, desc="Validation", leave=False):
            images    = images.to(device)
            meta_data = meta_data.to(device)
            labels    = labels.to(device)

            outputs = model(images, meta_data)
            loss    = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)
            preds = torch.argmax(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    avg_loss = running_loss / len(y_true)
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="macro", zero_division=0)
    rec  = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1   = f1_score(y_true, y_pred, average="macro", zero_division=0)
    spec = compute_specificity(y_true, y_pred, NUM_CLASSES)
    return avg_loss, acc, prec, rec, f1, spec, y_true, y_pred
```

```python
PHASE1_EPOCHS = 10

train_losses, val_losses = [], []
best_val_f1 = 0.0
best_model_state = None

for epoch in range(PHASE1_EPOCHS):
    print(f"\nPhase 1 Epoch [{epoch+1}/{PHASE1_EPOCHS}]")

    train_loss, train_acc, train_prec, train_rec, train_f1, train_spec = train_one_epoch(
        model, train_loader, optimizer, criterion, DEVICE
    )
    val_loss, val_acc, val_prec, val_rec, val_f1, val_spec, _, _ = evaluate(
        model, val_loader, criterion, DEVICE
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    scheduler.step()

    print(f"LR: {optimizer.param_groups[0]['lr']:.2e}")
    print(f"Train | Loss: {train_loss:.4f} | Acc: {train_acc:.4f} | Prec: {train_prec:.4f} | Rec: {train_rec:.4f} | F1: {train_f1:.4f} | Spec: {train_spec:.4f}")
    print(f"Val   | Loss: {val_loss:.4f} | Acc: {val_acc:.4f} | Prec: {val_prec:.4f} | Rec: {val_rec:.4f} | F1: {val_f1:.4f} | Spec: {val_spec:.4f}")

    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_model_state = model.state_dict()
        print(f"Validation F1 improved ({val_f1:.4f}) — saving best Phase 1 model")

if best_model_state is not None:
    model.load_state_dict(best_model_state)
    print("\nPhase 1 complete. Best model loaded.")
```

```python
BACKBONE_LR = 1e-6
HEAD_LR     = 1e-4

optimizer = torch.optim.AdamW([
    {"params": model.features.parameters(),    "lr": BACKBONE_LR},

    {"params": model.tabular_mlp.parameters(), "lr": HEAD_LR},
    {"params": model.meta_proj.parameters(),   "lr": HEAD_LR},
    {"params": model.classifier.parameters(),  "lr": HEAD_LR},
], weight_decay=5e-2)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=60, eta_min=1e-7
)
```

```python
PHASE2_EPOCHS = 60

fine_train_losses, fine_val_losses = [], []
best_val_f1 = 0.0
best_model_state = None

patience = 15
counter  = 0

for epoch in range(PHASE2_EPOCHS):
    print(f"\nPhase 2 Epoch [{epoch+1}/{PHASE2_EPOCHS}]")

    train_loss, train_acc, train_prec, train_rec, train_f1, train_spec = train_one_epoch(
        model, train_loader, optimizer, criterion, DEVICE
    )
    val_loss, val_acc, val_prec, val_rec, val_f1, val_spec, _, _ = evaluate(
        model, val_loader, criterion, DEVICE
    )

    fine_train_losses.append(train_loss)
    fine_val_losses.append(val_loss)
    scheduler.step()

    print(f"LR (backbone): {optimizer.param_groups[0]['lr']:.2e} | LR (head): {optimizer.param_groups[-1]['lr']:.2e}")
    print(f"Train | Loss: {train_loss:.4f} | Acc: {train_acc:.4f} | Prec: {train_prec:.4f} | Rec: {train_rec:.4f} | F1: {train_f1:.4f} | Spec: {train_spec:.4f}")
    print(f"Val   | Loss: {val_loss:.4f} | Acc: {val_acc:.4f} | Prec: {val_prec:.4f} | Rec: {val_rec:.4f} | F1: {val_f1:.4f} | Spec: {val_spec:.4f}")

    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_model_state = model.state_dict()
        torch.save(best_model_state, '/kaggle/working/ISIC2019_Ablation_C3_ImageMeta.pth')
        counter = 0
        print(f"Validation F1 improved ({val_f1:.4f}) — saving best Phase 2 model")
    else:
        counter += 1
        print(f"No improvement. Early stopping counter: {counter}/{patience}")

    if counter >= patience:
        print(f"\nEarly stopping triggered after {patience} epochs without F1 improvement.")
        break

if best_model_state is not None:
    model.load_state_dict(best_model_state)
    print("\nPhase 2 complete. Best model loaded.")
```

```python
from scipy import stats as scipy_stats

CONFIG = "C3"   # "C3" | "C4"

N_BOOTSTRAP = 1000
rng         = np.random.default_rng(42)
y_true_arr  = np.array(y_true)
y_pred_arr  = np.array(y_pred)
n_test      = len(y_true_arr)

boot = {"Accuracy": [], "F1 (Macro)": [], "Precision (Macro)": [], "Recall (Macro)": [], "Specificity": []}
per_class_f1 = {name: [] for name in class_names}

def compute_specificity_boot(y_t, y_p, num_classes):
    cm = confusion_matrix(y_t, y_p, labels=list(range(num_classes)))
    specs = []
    for i in range(num_classes):
        TP = cm[i, i]; FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP; TN = cm.sum() - (TP + FP + FN)
        specs.append(TN / (TN + FP) if (TN + FP) > 0 else 0.0)
    return float(np.mean(specs))

for _ in range(N_BOOTSTRAP):
    idx = rng.integers(0, n_test, size=n_test)
    y_t, y_p = y_true_arr[idx], y_pred_arr[idx]
    boot["Accuracy"].append(accuracy_score(y_t, y_p))
    boot["F1 (Macro)"].append(f1_score(y_t, y_p, average="macro", zero_division=0))
    boot["Precision (Macro)"].append(precision_score(y_t, y_p, average="macro", zero_division=0))
    boot["Recall (Macro)"].append(recall_score(y_t, y_p, average="macro", zero_division=0))
    boot["Specificity"].append(compute_specificity_boot(y_t, y_p, NUM_CLASSES))
    pc = f1_score(y_t, y_p, average=None, zero_division=0, labels=list(range(NUM_CLASSES)))
    for ci, name in enumerate(class_names):
        per_class_f1[name].append(pc[ci])

def ci95(arr): return float(np.percentile(arr, 2.5)), float(np.percentile(arr, 97.5))

short_names = [n.split("(")[-1].replace(")", "").strip() if "(" in n else n
               for n in class_names]

W = 70
print("\n" + "=" * W)
print(f"  ISIC 2019 — {CONFIG} — Bootstrap CI (n={N_BOOTSTRAP})")
print("=" * W)
print(f"{'Metric':<22} {'Mean':>7}  {'Std':>7}  {'CI Lower':>10}  {'CI Upper':>10}")
print("=" * W)
for metric, values in boot.items():
    arr = np.array(values); lo, hi = ci95(arr)
    print(f"{metric:<22} {arr.mean():>7.4f}  {arr.std():>7.4f}  {lo:>10.4f}  {hi:>10.4f}")
print("-" * W)
print(f"\n{'Class':<12} {'Mean':>7}  {'Std':>7}  {'CI Lower':>10}  {'CI Upper':>10}")
print("-" * W)
for sname, name in zip(short_names, class_names):
    arr = np.array(per_class_f1[name]); lo, hi = ci95(arr)
    print(f"{sname:<12} {arr.mean():>7.4f}  {arr.std():>7.4f}  {lo:>10.4f}  {hi:>10.4f}")
print("=" * W)

fig, axes = plt.subplots(1, 5, figsize=(18, 6))
fig.subplots_adjust(top=0.82, wspace=0.4)
for ax, (label, values) in zip(axes, boot.items()):
    arr = np.array(values); lo, hi = ci95(arr); mn, sd = arr.mean(), arr.std()
    ax.boxplot(arr, widths=0.5, patch_artist=True,
               boxprops=dict(facecolor="#4c72b0", alpha=0.75),
               medianprops=dict(color="white", linewidth=2),
               whiskerprops=dict(linewidth=1.2), capprops=dict(linewidth=1.2),
               flierprops=dict(marker="o", markersize=3, markerfacecolor="gray", alpha=0.5))
    ax.set_title(label, fontsize=10, fontweight="bold", pad=8)
    ax.set_xticks([]); ax.set_ylabel("Score", fontsize=9)
    ax.grid(axis="y", alpha=0.3, linestyle="--")
    ax.text(1, ax.get_ylim()[1], f"μ={mn:.4f}\nσ={sd:.4f}\n95%CI [{lo:.4f},{hi:.4f}]",
            ha="center", va="top", fontsize=7.5, color="#222222",
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightyellow", edgecolor="gray", alpha=0.8))
plt.suptitle(f"ISIC 2019 {CONFIG} — Bootstrap Metric Distributions (n={N_BOOTSTRAP}) — Test Set",
             fontsize=13, fontweight="bold", y=0.97)
plt.savefig(f"/kaggle/working/isic_{CONFIG.lower()}_bootstrap_boxplots.png", dpi=150, bbox_inches="tight")
plt.show()

means  = np.array([np.mean(per_class_f1[n]) for n in class_names])
stds   = np.array([np.std(per_class_f1[n])  for n in class_names])
lo_arr = np.array([ci95(per_class_f1[n])[0] for n in class_names])
hi_arr = np.array([ci95(per_class_f1[n])[1] for n in class_names])

fig, ax = plt.subplots(figsize=(13, 6))
x = np.arange(NUM_CLASSES)
bars = ax.bar(x, means, yerr=[means - lo_arr, hi_arr - means], capsize=6,
              color="#4c72b0", alpha=0.8, ecolor="#333333",
              error_kw={"linewidth": 1.8, "capthick": 1.8}, width=0.55)
ax.set_xticks(x)
ax.set_xticklabels(short_names, fontsize=11)
ax.set_ylabel("F1 Score", fontsize=11); ax.set_ylim(0, 1.15)
ax.set_title(f"ISIC 2019 {CONFIG} — Per-Class F1 Mean ± 95% CI (Bootstrap)",
             fontsize=13, fontweight="bold")
ax.grid(axis="y", alpha=0.3, linestyle="--"); ax.spines[["top", "right"]].set_visible(False)
for bar, m, s, hi in zip(bars, means, stds, hi_arr):
    ax.text(bar.get_x() + bar.get_width() / 2, hi + 0.025,
            f"{m:.3f}\n±{s:.3f}", ha="center", va="bottom", fontsize=9, color="#111111")
plt.tight_layout()
plt.savefig(f"/kaggle/working/isic_{CONFIG.lower()}_bootstrap_perclass_f1.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
def evaluate_and_plot_roc(model, loader, class_names):
    model.eval()
    all_targets, all_probs = [], []

    print("Extracting Softmax probabilities for ROC Curve...")
    with torch.no_grad():
        for images, meta_data, labels in loader:
            images    = images.to(DEVICE)
            meta_data = meta_data.to(DEVICE)
            probs     = F.softmax(model(images, meta_data), dim=1)
            all_targets.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    all_targets = np.array(all_targets)
    all_probs   = np.array(all_probs)
    y_true_bin  = label_binarize(all_targets, classes=range(len(class_names)))

    plt.figure(figsize=(12, 10))
    for i in range(len(class_names)):
        fpr, tpr, _ = roc_curve(y_true_bin[:, i], all_probs[:, i])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, lw=2, label=f"{class_names[i]} (AUC = {roc_auc:.3f})")

    plt.plot([0, 1], [0, 1], "k--", lw=2, label="Random Guessing")
    plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
    plt.xlabel("False Positive Rate", fontsize=12)
    plt.ylabel("True Positive Rate", fontsize=12)
    plt.title("ISIC 2019 ROC Curve -- C3 Image + Metadata", fontsize=16, fontweight="bold")
    plt.legend(loc="lower right", fontsize=10)
    plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

evaluate_and_plot_roc(model, test_loader, class_names)
```

```python
def evaluate_with_tta_and_thresholds(model, loader, device, thresholds):
    model.eval()
    y_true, y_pred = [], []

    with torch.no_grad():
        for images, meta_data, labels in tqdm(loader, desc="TTA Testing"):
            images    = images.to(device)
            meta_data = meta_data.to(device)

            out_orig  = F.softmax(model(images,                          meta_data), dim=1)
            out_hf    = F.softmax(model(torch.flip(images, [3]),         meta_data), dim=1)
            out_vf    = F.softmax(model(torch.flip(images, [2]),         meta_data), dim=1)
            out_r90   = F.softmax(model(torch.rot90(images, 1, [2, 3]), meta_data), dim=1)
            out_r270  = F.softmax(model(torch.rot90(images, 3, [2, 3]), meta_data), dim=1)

            avg_probs      = (out_orig + out_hf + out_vf + out_r90 + out_r270) / 5.0
            adjusted_probs = avg_probs / thresholds.to(device)
            preds          = torch.argmax(adjusted_probs, dim=1)

            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="macro", zero_division=0)
    rec  = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1   = f1_score(y_true, y_pred, average="macro", zero_division=0)
    spec = compute_specificity(y_true, y_pred, NUM_CLASSES)
    return acc, prec, rec, f1, spec, y_true, y_pred


custom_thresholds = torch.ones(NUM_CLASSES).to(DEVICE)
custom_thresholds[class_to_idx["NV"]]   = 1.0
custom_thresholds[class_to_idx["AK"]]   = 0.70
custom_thresholds[class_to_idx["SCC"]]  = 0.70
custom_thresholds[class_to_idx["DF"]]   = 0.75
custom_thresholds[class_to_idx["MEL"]]  = 0.90
custom_thresholds[class_to_idx["VASC"]] = 0.85
custom_thresholds[class_to_idx["BCC"]]  = 1.0
custom_thresholds[class_to_idx["BKL"]]  = 1.0

tta_acc, tta_prec, tta_rec, tta_f1, tta_spec, tta_true, tta_pred = evaluate_with_tta_and_thresholds(
    model, test_loader, DEVICE, custom_thresholds
)

print("\n--- FINAL TEST REPORT (C3 IMAGE + METADATA + TTA) ---")
print("Test Accuracy:", f"{tta_acc:.4f}")
print("Test Macro F1:", f"{tta_f1:.4f}")
print("\nTest Classification Report:\n")
print(classification_report(tta_true, tta_pred, target_names=class_names, zero_division=0, digits=4))

cm_tta = confusion_matrix(tta_true, tta_pred)
plt.figure(figsize=(11, 9))
sns.heatmap(cm_tta, annot=True, fmt="d", cmap="Oranges",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted (Refined Thresholds)"); plt.ylabel("Actual")
plt.title("ISIC 2019 Test Confusion Matrix -- C3 Image + Metadata (TTA)")
plt.tight_layout(); plt.show()
```

```python
def visualize_predictions(model, loader, class_names, num_images=10):
    model.eval()
    all_images, all_meta, all_labels = [], [], []
    for images, meta, labels in loader:
        all_images.append(images); all_meta.append(meta); all_labels.append(labels)
        if sum(len(x) for x in all_images) >= num_images * 2: break

    all_images = torch.cat(all_images)
    all_meta   = torch.cat(all_meta)
    all_labels = torch.cat(all_labels)
    indices    = random.sample(range(len(all_images)), min(num_images, len(all_images)))
    images     = all_images[indices].to(DEVICE)
    meta       = all_meta[indices].to(DEVICE)
    labels     = all_labels[indices]

    with torch.no_grad():
        preds = torch.argmax(model(images, meta), dim=1)

    mean = np.array([0.485, 0.456, 0.406])
    std  = np.array([0.229, 0.224, 0.225])
    cols = 5; rows = (num_images + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 3, rows * 3))
    axes = axes.flatten()

    for i in range(len(indices)):
        img = images[i].cpu().permute(1, 2, 0).numpy()
        img = np.clip(std * img + mean, 0, 1)
        correct = labels[i].item() == preds[i].item()
        axes[i].imshow(img)
        axes[i].set_title(
            f"T: {class_names[labels[i].item()].split('(')[0].strip()}\n"
            f"P: {class_names[preds[i].item()].split('(')[0].strip()}",
            fontsize=8, color="green" if correct else "red"
        )
        axes[i].axis("off")
    for j in range(len(indices), len(axes)):
        axes[j].axis("off")
    plt.suptitle("Test Predictions -- C3 Image + Metadata", fontsize=11)
    plt.tight_layout(); plt.show()

visualize_predictions(model, test_loader, class_names, num_images=10)
```
