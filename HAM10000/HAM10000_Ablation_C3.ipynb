```python
# HAM10000 — Ablation C3: Image + Metadata
# ConvNeXt-Tiny + Tabular MLP → concat → classifier
# No CLIP text, no dual backbone, no fusion mechanism.

import os, random
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

import numpy as np
import pandas as pd
from PIL import Image
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import transforms, models, datasets
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.amp import autocast, GradScaler

from torchinfo import summary
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.preprocessing import label_binarize
from sklearn.metrics import (
    classification_report, confusion_matrix,
    f1_score, precision_score, recall_score,
    accuracy_score, roc_curve, auc
)

import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark     = False
os.environ["PYTHONHASHSEED"]       = str(SEED)
```

```python
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 16

TRAIN_DIR = "/kaggle/input/datasets/sujansarkar/ham10000-folder-wise/ham10000_80_20/ham10000_80_20/train_dir"
TEST_DIR  = "/kaggle/input/datasets/sujansarkar/ham10000-folder-wise/ham10000_80_20/ham10000_80_20/test_dir"
META_CSV  = "/kaggle/input/datasets/sujansarkar/ham10000-folder-wise/HAM10000_metadata.csv"

train_tfms = transforms.Compose([
    transforms.Resize(288),
    transforms.RandomResizedCrop(256, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomApply(
        [transforms.ColorJitter(brightness=0.2, contrast=0.2, hue=0.05)],
        p=0.5
    ),
    transforms.RandomAffine(
        degrees=180, translate=(0.1, 0.1), scale=(0.9, 1.1), shear=15
    ),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

val_test_tfms = transforms.Compose([
    transforms.Resize(288),
    transforms.CenterCrop(256),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

print(f"Device: {DEVICE}")
```

```python
full_train_folder = datasets.ImageFolder(TRAIN_DIR)
test_folder       = datasets.ImageFolder(TEST_DIR)

all_paths  = [p for p, _ in full_train_folder.imgs]
all_labels = [l for _, l in full_train_folder.imgs]

class_to_idx = full_train_folder.class_to_idx
idx_to_class = {v: k for k, v in class_to_idx.items()}
NUM_CLASSES  = len(class_to_idx)

# Same 80/20 split used across all C2–C6 for consistency
sss = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, val_idx = next(sss.split(all_paths, all_labels))

train_paths  = [all_paths[i]  for i in train_idx]
train_labels = [all_labels[i] for i in train_idx]
val_paths    = [all_paths[i]  for i in val_idx]
val_labels   = [all_labels[i] for i in val_idx]

test_paths  = [p for p, _ in test_folder.imgs]
test_labels = [l for _, l in test_folder.imgs]

print(f"Train: {len(train_paths)} | Val: {len(val_paths)} | Test: {len(test_paths)}")
print(f"Classes ({NUM_CLASSES}): {class_to_idx}")
```

```python
NUMERICAL_COLS   = ['age']
CATEGORICAL_COLS = ['sex', 'localization']

meta_df = pd.read_csv(META_CSV).set_index('image_id')

# Compute normalisation stats from training split only — no leakage
train_ids   = [os.path.splitext(os.path.basename(p))[0] for p in train_paths]
train_meta  = meta_df.loc[meta_df.index.intersection(train_ids)]
age_mean    = train_meta['age'].mean()
age_std     = train_meta['age'].std() + 1e-6

# Apply normalisation to the full dataframe
meta_df['age'] = meta_df['age'].fillna(age_mean)
meta_df['age'] = (meta_df['age'] - age_mean) / age_std

# Fill categorical missing values
meta_df['sex']          = meta_df['sex'].fillna('unknown')
meta_df['localization'] = meta_df['localization'].fillna('unknown')

# Build one-hot schema from training split only
train_meta_norm = meta_df.loc[meta_df.index.intersection(train_ids)]
cat_dummies     = pd.get_dummies(
    train_meta_norm[CATEGORICAL_COLS], prefix=CATEGORICAL_COLS
)
CAT_COLUMNS = cat_dummies.columns.tolist()
NUM_META    = len(NUMERICAL_COLS) + len(CAT_COLUMNS)

def get_meta_vector(image_id):
    """Return a normalised metadata tensor for a given image_id."""
    if image_id not in meta_df.index:
        return torch.zeros(NUM_META, dtype=torch.float32)

    row = meta_df.loc[image_id]
    num = [float(row['age'])]

    cat = pd.get_dummies(
        pd.DataFrame([row[CATEGORICAL_COLS].to_dict()]),
        prefix=CATEGORICAL_COLS
    ).reindex(columns=CAT_COLUMNS, fill_value=0).values[0].tolist()

    return torch.tensor(num + cat, dtype=torch.float32)

print(f"Metadata feature dimensionality: {NUM_META}")
print(f"Numerical : {NUMERICAL_COLS}")
print(f"Categorical columns after encoding: {CAT_COLUMNS}")
```

```python
class HAMImageMetaDataset(Dataset):
    """
    Image + Metadata dataset for HAM10000.
    Each sample returns (image, tabular_vector, label).
    class_to_idx is fixed from the full training ImageFolder.
    Metadata is matched by image_id (filename without extension).
    """
    def __init__(self, img_paths, labels, transform=None):
        self.img_paths = img_paths
        self.labels    = labels
        self.transform = transform

    def __len__(self):
        return len(self.img_paths)

    def __getitem__(self, idx):
        path     = self.img_paths[idx]
        image_id = os.path.splitext(os.path.basename(path))[0]

        image = Image.open(path).convert('RGB')
        if self.transform:
            image = self.transform(image)

        tab   = get_meta_vector(image_id)
        label = self.labels[idx]

        return image, tab, label


train_dataset = HAMImageMetaDataset(train_paths, train_labels, train_tfms)
val_dataset   = HAMImageMetaDataset(val_paths,   val_labels,   val_test_tfms)
test_dataset  = HAMImageMetaDataset(test_paths,  test_labels,  val_test_tfms)
```

```python
class_counts         = Counter(train_labels)
total_samples        = sum(class_counts.values())
raw_weights          = [
    total_samples / (NUM_CLASSES * class_counts[i] ** 0.5)
    for i in range(NUM_CLASSES)
]
mean_w               = sum(raw_weights) / NUM_CLASSES
class_weights_tensor = torch.FloatTensor([w / mean_w for w in raw_weights])


def seed_worker(worker_id):
    seed = SEED + worker_id
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=4, pin_memory=True, drop_last=True,
    persistent_workers=True, worker_init_fn=seed_worker,
    generator=torch.Generator().manual_seed(SEED)
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=4, pin_memory=True, persistent_workers=True,
    worker_init_fn=seed_worker
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=4, pin_memory=True, persistent_workers=True,
    worker_init_fn=seed_worker
)

print(f"Train: {len(train_dataset)} | Val: {len(val_dataset)} | Test: {len(test_dataset)}")
```

```python
IDX_TO_CLASS = {
    class_to_idx["akiec"]: "Actinic Keratoses / IECSC (AKIEC)",
    class_to_idx["bcc"]  : "Basal Cell Carcinoma (BCC)",
    class_to_idx["bkl"]  : "Benign Keratosis-like Lesions (BKL)",
    class_to_idx["df"]   : "Dermatofibroma (DF)",
    class_to_idx["mel"]  : "Melanoma (MEL)",
    class_to_idx["nv"]   : "Melanocytic Nevi (NV)",
    class_to_idx["vasc"] : "Vascular Lesions (VASC)",
}
class_names = [IDX_TO_CLASS[i] for i in range(NUM_CLASSES)]

print(f"Classes ({NUM_CLASSES}):", class_names)
print("\nFocalLoss weights:", {
    IDX_TO_CLASS[i]: f"{class_weights_tensor[i]:.3f}"
    for i in range(NUM_CLASSES)
})
print("\nTrain class counts:", {
    IDX_TO_CLASS[k]: v for k, v in class_counts.items()
})
```

```python
class ConvNeXtWithMeta(nn.Module):
    """
    HAM10000 Ablation C3 -- ConvNeXt-Tiny + Tabular Metadata.
    Image features from ConvNeXt-Tiny and encoded metadata are
    concatenated and passed to a shared classifier head.
    No Swin-Tiny, no CLIP text encoder, no cross-attention, no gate.
    Isolates the contribution of patient metadata over image-only (C2).
    """
    def __init__(self, num_meta_features, num_classes=7,
                 pretrained=True, freeze_backbone=False):
        super().__init__()

        # ConvNeXt-Tiny vision branch
        cnxt              = models.convnext_tiny(
            weights=models.ConvNeXt_Tiny_Weights.DEFAULT if pretrained else None
        )
        self.features     = cnxt.features
        self.pool         = nn.AdaptiveAvgPool2d(1)
        self.vision_dim   = 768

        # Tabular MLP
        self.tabular_mlp = nn.Sequential(
            nn.Linear(num_meta_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3)
        )
        self.tabular_dim = 128

        # Classifier on concatenated features [B, 768 + 128] = [B, 896]
        fused_dim = self.vision_dim + self.tabular_dim
        self.classifier = nn.Sequential(
            nn.Dropout(0.4),
            nn.Linear(fused_dim, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(512, num_classes)
        )

        if freeze_backbone:
            for p in self.features.parameters():
                p.requires_grad = False

    def forward(self, x, tab):
        # Vision: [B, 768, 8, 8] → pool → [B, 768]
        f_vis = self.pool(self.features(x)).flatten(1)

        # Metadata: [B, d] → [B, 128]
        f_tab = self.tabular_mlp(tab)

        # Concat and classify
        fused = torch.cat([f_vis, f_tab], dim=1)   # [B, 896]
        return self.classifier(fused)


model = ConvNeXtWithMeta(
    num_meta_features=NUM_META,
    num_classes=NUM_CLASSES,
    pretrained=True,
    freeze_backbone=True
).to(DEVICE)

summary(
    model,
    input_size=[(BATCH_SIZE, 3, 256, 256), (BATCH_SIZE, NUM_META)],
    dtypes=[torch.float32, torch.float32],
    col_names=["input_size", "output_size", "num_params"],
    depth=3
)
```

```python
class FocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=2.0, label_smoothing=0.1):
        super().__init__()
        self.gamma = gamma
        self.ce    = nn.CrossEntropyLoss(
            weight=weight,
            label_smoothing=label_smoothing,
            reduction="none"
        )

    def forward(self, inputs, targets):
        ce_loss    = self.ce(inputs, targets)
        pt         = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean()


criterion = FocalLoss(
    weight=class_weights_tensor.to(DEVICE),
    gamma=2.0,
    label_smoothing=0.1
)
scaler = GradScaler("cuda")
```

```python
def compute_specificity(y_true, y_pred, num_classes):
    cm    = confusion_matrix(y_true, y_pred, labels=range(num_classes))
    specs = []
    for i in range(num_classes):
        TP = cm[i, i]
        FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP
        TN = cm.sum() - (TP + FP + FN)
        specs.append(TN / (TN + FP) if (TN + FP) > 0 else 0.0)
    return float(np.mean(specs))


def mixup_batch(images, tab, labels, alpha=0.3):
    lam      = np.random.beta(alpha, alpha)
    rand_idx = torch.randperm(images.size(0), device=images.device)
    mix_img  = lam * images + (1 - lam) * images[rand_idx]
    mix_tab  = lam * tab    + (1 - lam) * tab[rand_idx]
    return mix_img, mix_tab, labels, labels[rand_idx], lam


def train_one_epoch(model, loader, optimizer, criterion, device,
                    use_mixup=True):
    model.train()
    running_loss = 0.0
    y_true, y_pred = [], []

    for images, tab, labels in tqdm(loader, desc="Training", leave=False):
        images, tab, labels = (images.to(device),
                               tab.to(device),
                               labels.to(device))

        if use_mixup:
            mix_img, mix_tab, y_a, y_b, lam = mixup_batch(
                images, tab, labels
            )
        else:
            mix_img, mix_tab, y_a, y_b, lam = images, tab, labels, labels, 1.0

        optimizer.zero_grad()
        with autocast("cuda"):
            out  = model(mix_img, mix_tab)
            loss = (lam * criterion(out, y_a) +
                    (1 - lam) * criterion(out, y_b))

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item() * images.size(0)
        y_true.extend((y_a if lam >= 0.5 else y_b).cpu().numpy())
        y_pred.extend(torch.argmax(out, 1).cpu().numpy())

    n = len(y_true)
    return (
        running_loss / n,
        accuracy_score(y_true, y_pred),
        precision_score(y_true, y_pred, average="macro", zero_division=0),
        recall_score(y_true, y_pred,    average="macro", zero_division=0),
        f1_score(y_true, y_pred,        average="macro", zero_division=0),
        compute_specificity(y_true, y_pred, NUM_CLASSES)
    )


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    y_true, y_pred = [], []

    with torch.no_grad():
        for images, tab, labels in tqdm(loader, desc="Eval", leave=False):
            images, tab, labels = (images.to(device),
                                   tab.to(device),
                                   labels.to(device))
            out  = model(images, tab)
            loss = criterion(out, labels)
            running_loss += loss.item() * images.size(0)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(torch.argmax(out, 1).cpu().numpy())

    n = len(y_true)
    return (
        running_loss / n,
        accuracy_score(y_true, y_pred),
        precision_score(y_true, y_pred, average="macro", zero_division=0),
        recall_score(y_true, y_pred,    average="macro", zero_division=0),
        f1_score(y_true, y_pred,        average="macro", zero_division=0),
        compute_specificity(y_true, y_pred, NUM_CLASSES),
        y_true, y_pred
    )
```

```python
PHASE1_EPOCHS = 20

optimizer_p1 = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4, weight_decay=1e-2
)
scheduler_p1 = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
    optimizer_p1, T_0=10, T_mult=2, eta_min=1e-6
)

train_losses, val_losses = [], []
best_val_f1, best_model_state = 0.0, None

print("=" * 60)
print("PHASE 1 — Frozen backbone | HAM10000 Model 2 C3")
print("=" * 60)

for epoch in range(PHASE1_EPOCHS):
    print(f"\nPhase 1 Epoch [{epoch+1}/{PHASE1_EPOCHS}]")
    tr = train_one_epoch(model, train_loader, optimizer_p1, criterion, DEVICE)
    vl = evaluate(model, val_loader, criterion, DEVICE)
    train_losses.append(tr[0]); val_losses.append(vl[0])
    scheduler_p1.step()

    print(f"LR: {optimizer_p1.param_groups[0]['lr']:.2e}")
    print(f"Train | Loss:{tr[0]:.4f} Acc:{tr[1]:.4f} "
          f"Prec:{tr[2]:.4f} Rec:{tr[3]:.4f} F1:{tr[4]:.4f} Spec:{tr[5]:.4f}")
    print(f"Val   | Loss:{vl[0]:.4f} Acc:{vl[1]:.4f} "
          f"Prec:{vl[2]:.4f} Rec:{vl[3]:.4f} F1:{vl[4]:.4f} Spec:{vl[5]:.4f}")

    if vl[4] > best_val_f1:
        best_val_f1      = vl[4]
        best_model_state = model.state_dict()
        print(f"  Val F1 improved ({best_val_f1:.4f}) — checkpoint saved")

model.load_state_dict(best_model_state)
print(f"\nPhase 1 complete. Best val F1: {best_val_f1:.4f}")
```

```python
for p in model.features.parameters():
    p.requires_grad = True

PHASE2_EPOCHS = 60
BACKBONE_LR   = 1e-6
HEAD_LR       = 1e-4

optimizer_p2 = torch.optim.AdamW([
    {"params": model.features.parameters(),     "lr": BACKBONE_LR},
    {"params": model.tabular_mlp.parameters(),  "lr": HEAD_LR},
    {"params": model.classifier.parameters(),   "lr": HEAD_LR},
], weight_decay=5e-2)

scheduler_p2 = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
    optimizer_p2, T_0=10, T_mult=2, eta_min=1e-7
)

fine_train_losses, fine_val_losses = [], []
best_val_f1, best_model_state = 0.0, None
patience, counter = 15, 0

print("=" * 60)
print("PHASE 2 — Full fine-tuning | HAM10000 Model 2 C3")
print(f"Backbone LR: {BACKBONE_LR:.0e} | Head LR: {HEAD_LR:.0e}")
print("=" * 60)

for epoch in range(PHASE2_EPOCHS):
    print(f"\nPhase 2 Epoch [{epoch+1}/{PHASE2_EPOCHS}]")
    tr = train_one_epoch(model, train_loader, optimizer_p2, criterion, DEVICE)
    vl = evaluate(model, val_loader, criterion, DEVICE)
    fine_train_losses.append(tr[0]); fine_val_losses.append(vl[0])
    scheduler_p2.step()

    print(f"LR (backbone):{optimizer_p2.param_groups[0]['lr']:.2e} | "
          f"LR (head):{optimizer_p2.param_groups[1]['lr']:.2e}")
    print(f"Train | Loss:{tr[0]:.4f} Acc:{tr[1]:.4f} "
          f"Prec:{tr[2]:.4f} Rec:{tr[3]:.4f} F1:{tr[4]:.4f} Spec:{tr[5]:.4f}")
    print(f"Val   | Loss:{vl[0]:.4f} Acc:{vl[1]:.4f} "
          f"Prec:{vl[2]:.4f} Rec:{vl[3]:.4f} F1:{vl[4]:.4f} Spec:{vl[5]:.4f}")

    if vl[4] > best_val_f1:
        best_val_f1      = vl[4]
        best_model_state = model.state_dict()
        torch.save(best_model_state,
                   "/kaggle/working/HAM10000_Model2_C3_ConvNeXtMeta.pth")
        counter = 0
        print(f"  Val F1 improved ({best_val_f1:.4f}) — saved")
    else:
        counter += 1
        print(f"  No improvement. Counter: {counter}/{patience}")

    if counter >= patience:
        print("\nEarly stopping triggered.")
        break

model.load_state_dict(best_model_state)
print(f"\nPhase 2 complete. Best val F1: {best_val_f1:.4f}")
```

```python
unweighted_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=4, pin_memory=True, persistent_workers=True,
    worker_init_fn=seed_worker
)

for pg in optimizer_p2.param_groups:
    pg['lr'] = 1e-6

PHASE3_EPOCHS = 10
calib_train_losses, calib_val_losses = [], []

print("=" * 60)
print("PHASE 3 — Calibration (LR locked 1e-6, no MixUp)")
print("=" * 60)

for epoch in range(PHASE3_EPOCHS):
    print(f"\nPhase 3 Epoch [{epoch+1}/{PHASE3_EPOCHS}]")
    tr = train_one_epoch(
        model, unweighted_loader, optimizer_p2,
        criterion, DEVICE, use_mixup=False
    )
    vl = evaluate(model, val_loader, criterion, DEVICE)
    calib_train_losses.append(tr[0]); calib_val_losses.append(vl[0])

    print(f"Train | Loss:{tr[0]:.4f} Acc:{tr[1]:.4f} F1:{tr[4]:.4f}")
    print(f"Val   | Loss:{vl[0]:.4f} Acc:{vl[1]:.4f} F1:{vl[4]:.4f}")

    if vl[4] > best_val_f1:
        best_val_f1      = vl[4]
        best_model_state = model.state_dict()
        print(f"  Phase 3 val F1 improved ({best_val_f1:.4f}) — saved")

if best_model_state:
    model.load_state_dict(best_model_state)
print(f"\nBest overall val F1: {best_val_f1:.4f}")
```

```python
def evaluate_with_tta_and_thresholds(model, loader, device, thresholds):
    model.eval()
    all_probs, y_true = [], []

    with torch.no_grad():
        for images, tab, labels in tqdm(loader, desc="TTA", leave=False):
            images, tab = images.to(device), tab.to(device)

            def fwd(img):
                return F.softmax(model(img, tab), dim=1)

            avg = (fwd(images) +
                   fwd(torch.flip(images, [3])) +
                   fwd(torch.flip(images, [2])) +
                   fwd(torch.rot90(images, 1, [2, 3])) +
                   fwd(torch.rot90(images, 3, [2, 3]))) / 5.0

            all_probs.extend(avg.cpu().numpy())
            y_true.extend(labels.cpu().numpy())

    all_probs = np.array(all_probs)
    thresh_np = thresholds.cpu().numpy()
    y_pred    = np.argmax(all_probs / thresh_np, axis=1)

    return (
        accuracy_score(y_true, y_pred),
        precision_score(y_true, y_pred, average="macro", zero_division=0),
        recall_score(y_true,   y_pred,  average="macro", zero_division=0),
        f1_score(y_true,       y_pred,  average="macro", zero_division=0),
        compute_specificity(y_true, y_pred, NUM_CLASSES),
        y_true, y_pred, all_probs
    )


custom_thresholds = torch.ones(NUM_CLASSES)
custom_thresholds[class_to_idx["mel"]]   = 0.75
custom_thresholds[class_to_idx["akiec"]] = 0.80
custom_thresholds[class_to_idx["bcc"]]   = 0.85
custom_thresholds[class_to_idx["bkl"]]   = 1.0
custom_thresholds[class_to_idx["df"]]    = 1.0
custom_thresholds[class_to_idx["nv"]]    = 1.0
custom_thresholds[class_to_idx["vasc"]]  = 1.0

acc, prec, rec, f1, spec, tta_true, tta_pred, tta_probs = \
    evaluate_with_tta_and_thresholds(
        model, test_loader, DEVICE, custom_thresholds
    )

print("\n--- FINAL TEST (ConvNeXt + Metadata + TTA + Thresholds) ---")
print(f"Acc:{acc:.4f} | Prec:{prec:.4f} | Rec:{rec:.4f} "
      f"| F1:{f1:.4f} | Spec:{spec:.4f}")
print("\n" + classification_report(
    tta_true, tta_pred, target_names=class_names, zero_division=0, digits=4
))

cm_tta = confusion_matrix(tta_true, tta_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm_tta, annot=True, fmt="d", cmap="Oranges",
            xticklabels=class_names, yticklabels=class_names, square=True)
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.title("HAM10000 Test CM — Model 2 C3 + TTA + Thresholds")
plt.tight_layout(); plt.show()

y_bin = label_binarize(tta_true, classes=range(NUM_CLASSES))
plt.figure(figsize=(10, 8))
for i in range(NUM_CLASSES):
    fpr, tpr, _ = roc_curve(y_bin[:, i], tta_probs[:, i])
    plt.plot(fpr, tpr, lw=2,
             label=f"{class_names[i]} (AUC={auc(fpr, tpr):.3f})")
plt.plot([0,1],[0,1], "k--", lw=2, label="Random Guessing")
plt.xlim([0,1]); plt.ylim([0,1.05])
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.title("HAM10000 ROC — Model 2 C3: ConvNeXt-Tiny + Metadata",
          fontsize=13, fontweight="bold")
plt.legend(loc="lower right", fontsize=8)
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```
