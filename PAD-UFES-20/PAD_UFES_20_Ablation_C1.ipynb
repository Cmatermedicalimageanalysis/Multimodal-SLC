```python
#PAD-UFES-20-C1
import os
import numpy as np
import pandas as pd
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import label_binarize, StandardScaler
from sklearn.metrics import roc_curve, auc
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score
)

import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 128
meta_path = "/kaggle/input/datasets/sujansarkar/pad-ufes-20-classification/metadata.csv"
df = pd.read_csv(meta_path)

gss_train = GroupShuffleSplit(n_splits=1, train_size=0.7, random_state=42)
train_idx, temp_idx = next(gss_train.split(df, groups=df['patient_id']))

train_df = df.iloc[train_idx].copy()
temp_df = df.iloc[temp_idx].copy()

gss_val = GroupShuffleSplit(n_splits=1, train_size=0.5, random_state=42)
val_idx, test_idx = next(gss_val.split(temp_df, groups=temp_df['patient_id']))

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()

print("--- PAD-UFES-20 Data Split Report ---")
print(f"Train Set: {len(train_df):>4} images | Patients: {train_df['patient_id'].nunique()}")
print(f"Val Set:   {len(val_df):>4} images | Patients: {val_df['patient_id'].nunique()}")
print(f"Test Set:  {len(test_df):>4} images | Patients: {test_df['patient_id'].nunique()}")
```

```python
NUMERICAL_COLS = ['age', 'diameter_1', 'diameter_2']
CATEGORICAL_COLS = [
    'gender', 'fitspatrick', 'region', 'elevation', 
    'smoke', 'drink', 'pesticide', 'skin_cancer_history', 'cancer_history',
    'has_piped_water', 'has_sewage_system', 'itch', 'grew', 'hurt', 'changed', 'bleed',
    'background_father', 'background_mother'
]

def encode_categoricals(dataframe):
    """Handles type coercion and one-hot encoding only.
    Numerical normalization is done separately using train-only statistics."""
    temp_df = dataframe.copy()

    for col in NUMERICAL_COLS:
        temp_df[col] = pd.to_numeric(temp_df[col], errors='coerce')

    for col in CATEGORICAL_COLS:
        temp_df[col] = temp_df[col].fillna('UNKNOWN').astype(str)

    # One-Hot Encoding
    temp_df = pd.get_dummies(temp_df, columns=CATEGORICAL_COLS, dummy_na=False)

    if 'img_id' in temp_df.columns:
        temp_df.set_index('img_id', inplace=True)
    return temp_df

train_df = encode_categoricals(train_df)
val_df   = encode_categoricals(val_df)
test_df  = encode_categoricals(test_df)

# Align columns so the network doesn't crash if a category is missing in val/test set
train_df, val_df  = train_df.align(val_df,  join='left', axis=1, fill_value=0)
train_df, test_df = train_df.align(test_df, join='left', axis=1, fill_value=0)

# ── FIX: impute NaN using train medians only, then fit scaler on train only ──
# Using train statistics for val/test prevents distribution leakage.
train_num_medians = {col: train_df[col].median() for col in NUMERICAL_COLS}
for col in NUMERICAL_COLS:
    train_df[col] = train_df[col].fillna(train_num_medians[col])
    val_df[col]   = val_df[col].fillna(train_num_medians[col])
    test_df[col]  = test_df[col].fillna(train_num_medians[col])

scalers = {}
for col in NUMERICAL_COLS:
    sc = StandardScaler()
    train_df[f"{col}_norm"] = sc.fit_transform(train_df[[col]]).ravel()
    val_df[f"{col}_norm"]   = sc.transform(val_df[[col]]).ravel()
    test_df[f"{col}_norm"]  = sc.transform(test_df[[col]]).ravel()
    scalers[col] = sc

META_COLS = [c for c in train_df.columns if c.startswith(tuple(CATEGORICAL_COLS)) or c.endswith('_norm')]
NUM_META_FEATURES = len(META_COLS)

print(f"Tabular features processed. MLP Input Vector Size: {NUM_META_FEATURES}")
```

```python
class TabularOnlyDataset(Dataset):
    def __init__(self, df, meta_cols):
        self.df = df
        self.meta_cols = meta_cols
        self.classes = sorted(df['diagnostic'].unique())
        self.class_to_idx = {cls_name: i for i, cls_name in enumerate(self.classes)}

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        meta_data = row[self.meta_cols].values.astype(np.float32)
        label_str = row['diagnostic']
        label = self.class_to_idx[label_str]
        return torch.tensor(meta_data), torch.tensor(label)

train_dataset = TabularOnlyDataset(train_df, META_COLS)
val_dataset   = TabularOnlyDataset(val_df,   META_COLS)
test_dataset  = TabularOnlyDataset(test_df,  META_COLS)

# ── FIX: use class weights in FocalLoss instead of WeightedRandomSampler ─────
# Passing weights to the loss is the canonical approach when using FocalLoss;
# using both a sampler and a weighted loss double-counts the imbalance.
train_targets = [train_dataset.class_to_idx[label] for label in train_df['diagnostic']]
class_counts  = Counter(train_targets)
NUM_CLASSES   = len(class_counts)
total_samples = sum(class_counts.values())

# Inverse-frequency weights, sqrt-smoothed for stability
class_weight_list  = [total_samples / (NUM_CLASSES * class_counts[i] ** 0.5) for i in range(NUM_CLASSES)]
# Normalise so weights average to 1.0
mean_w = sum(class_weight_list) / NUM_CLASSES
class_weights_tensor = torch.FloatTensor([w / mean_w for w in class_weight_list])

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False)

# Mappings
class_to_idx = train_dataset.class_to_idx
class_names  = [cls for cls in class_to_idx.keys()]
print(f"Classes: {class_names}")
print(f"Class weights (normalised): { {class_names[i]: f'{class_weights_tensor[i]:.3f}' for i in range(NUM_CLASSES)} }")
```

```python
class TabularNet(nn.Module):
    def __init__(self, num_meta_features, num_classes=6):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(num_meta_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            
            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            
            nn.Linear(64, num_classes)
        )

    def forward(self, x_meta):
        return self.mlp(x_meta)

model = TabularNet(num_meta_features=NUM_META_FEATURES, num_classes=NUM_CLASSES).to(DEVICE)
```

```python
class FocalLoss(nn.Module):
    def __init__(self, weight=None, gamma=2.0):
        super().__init__()
        self.gamma = gamma
        # ── FIX: removed label_smoothing — it conflicts with focal loss ──────
        # Focal loss already handles hard/minority examples via the (1-pt)^gamma
        # modulating factor. Label smoothing softens targets in the opposite
        # direction, partially cancelling the focal effect.
        self.ce = nn.CrossEntropyLoss(weight=weight, reduction="none")

    def forward(self, inputs, targets):
        ce_loss = self.ce(inputs, targets)
        pt = torch.exp(-ce_loss)
        focal_loss = ((1 - pt) ** self.gamma) * ce_loss
        return focal_loss.mean()
```

```python
# ── FIX: class weights are now passed to FocalLoss (not to a sampler) ───────
criterion  = FocalLoss(weight=class_weights_tensor.to(DEVICE), gamma=2.0).to(DEVICE)
optimizer  = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-2)
scheduler  = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=5)
```

```python
def compute_specificity(y_true, y_pred, num_classes):
    cm = confusion_matrix(y_true, y_pred)
    specificities = []
    for i in range(num_classes):
        TP = cm[i, i]
        FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP
        TN = cm.sum() - (TP + FP + FN)
        specificities.append(TN / (TN + FP) if (TN + FP) > 0 else 0)
    return np.mean(specificities)

def train_tabular(model, loader, optimizer, criterion, device):
    model.train()
    running_loss, y_true, y_pred = 0.0, [], []

    for meta_data, labels in loader:
        meta_data, labels = meta_data.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(meta_data)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * meta_data.size(0)
        preds = torch.argmax(outputs, 1)
        
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())

    return running_loss / len(y_true), accuracy_score(y_true, y_pred), f1_score(y_true, y_pred, average="macro", zero_division=0)

def evaluate_tabular(model, loader, criterion, device):
    model.eval()
    running_loss, y_true, y_pred = 0.0, [], []

    with torch.no_grad():
        for meta_data, labels in loader:
            meta_data, labels = meta_data.to(device), labels.to(device)
            outputs = model(meta_data)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item() * meta_data.size(0)
            preds = torch.argmax(outputs, 1)

            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    avg_loss = running_loss / len(y_true)
    acc = accuracy_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)
    spec = compute_specificity(y_true, y_pred, NUM_CLASSES)
    return avg_loss, acc, f1, spec, y_true, y_pred
```

```python
from tqdm import tqdm

EPOCHS = 80
best_val_f1 = 0.0
best_model_state = None
best_f1_epoch = 1  # FIX: track the epoch that gave best val F1 for plot annotation

# Added lists to track history for your loss curves!
train_losses = []
val_losses   = []

print("Starting Tabular-Only Ablation Study...")

# Wrap the epoch range with tqdm
epoch_pbar = tqdm(range(EPOCHS), desc="Training Progress")

for epoch in epoch_pbar:
    train_loss, train_acc, train_f1 = train_tabular(model, train_loader, optimizer, criterion, DEVICE)
    val_loss, val_acc, val_f1, val_spec, _, _ = evaluate_tabular(model, val_loader, criterion, DEVICE)
    
    train_losses.append(train_loss)
    val_losses.append(val_loss)
    
    scheduler.step(val_f1)
    
    # Update the progress bar with the latest metrics
    epoch_pbar.set_postfix({
        'Train F1': f'{train_f1:.4f}', 
        'Val F1': f'{val_f1:.4f}', 
        'Val Acc': f'{val_acc:.4f}', 
        'Val Spec': f'{val_spec:.4f}'
    })
    
    if val_f1 > best_val_f1:
        best_val_f1   = val_f1
        best_f1_epoch = epoch + 1   # 1-indexed for the plot
        best_model_state = model.state_dict()

# Load best for final testing
if best_model_state:
    model.load_state_dict(best_model_state)
    print("\nLoaded best model state based on Validation F1.")
```

```python
def print_full_report(split_name, model, loader, criterion, device):
    loss, acc, f1, spec, y_true, y_pred = evaluate_tabular(model, loader, criterion, device)
    
    print(f"\n{'='*40}")
    print(f"{split_name.upper()} REPORT")
    print(f"{'='*40}")
    print(f"{split_name} Loss: {loss:.4f}")
    print(f"{split_name} Accuracy: {acc:.4f}")
    print(f"{split_name} Specificity (Macro): {spec:.4f}")

    print(f"\n{split_name} Classification Report:\n")
    print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0, digits=4))

    macro_prec = precision_score(y_true, y_pred, average="macro", zero_division=0)
    macro_rec  = recall_score(y_true, y_pred, average="macro", zero_division=0)
    
    weighted_prec = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    weighted_rec  = recall_score(y_true, y_pred, average="weighted", zero_division=0)
    weighted_f1   = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    print(f"{split_name} Additional Summary:")
    print(f"Macro Avg     -> Precision: {macro_prec:.4f} | Recall: {macro_rec:.4f} | F1: {f1:.4f}")
    print(f"Weighted Avg  -> Precision: {weighted_prec:.4f} | Recall: {weighted_rec:.4f} | F1: {weighted_f1:.4f}")
    print(f"Specificity   -> {spec:.4f}\n")
    
    return y_true, y_pred

_, _ = print_full_report("Train", model, train_loader, criterion, DEVICE)
_, _ = print_full_report("Validation", model, val_loader, criterion, DEVICE)
y_true_test, y_pred_test = print_full_report("Test", model, test_loader, criterion, DEVICE)

cm = confusion_matrix(y_true_test, y_pred_test)

plt.figure(figsize=(10, 8))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="mako_r", vmax=100,
    xticklabels=class_names, yticklabels=class_names,
    square=True, cbar_kws={"shrink": 0.8}
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Tabular-Only Test Confusion Matrix")
plt.tight_layout()
plt.show()

model.eval()
all_targets, all_probs = [], []
with torch.no_grad():
    for meta_data, labels in test_loader:
        meta_data = meta_data.to(DEVICE)
        probs = F.softmax(model(meta_data), dim=1)
        all_targets.extend(labels.numpy())
        all_probs.extend(probs.cpu().numpy())

y_true_bin = label_binarize(all_targets, classes=range(NUM_CLASSES))
all_probs = np.array(all_probs)

plt.figure(figsize=(10, 8))
for i in range(NUM_CLASSES):
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], all_probs[:, i])
    plt.plot(fpr, tpr, lw=2, label=f'{class_names[i]} (AUC = {auc(fpr, tpr):.3f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Guessing')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.title('Tabular-Only Multi-Class ROC Curve', fontsize=16, fontweight='bold')
plt.legend(loc="lower right", fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

epochs_range = range(1, len(train_losses) + 1)

plt.figure(figsize=(10, 5))
plt.plot(epochs_range, train_losses, label="Training Loss")
plt.plot(epochs_range, val_losses, label="Validation Loss")
# FIX: annotate the epoch where val F1 was best (consistent with checkpoint criterion)
plt.axvline(best_f1_epoch, color="green", linestyle="--", linewidth=1.2, label=f"Best Val F1 Epoch ({best_f1_epoch})")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss (Tabular-Only Ablation)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from scipy import stats as scipy_stats

N_BOOTSTRAP = 1000
rng        = np.random.default_rng(42)
y_true_arr = np.array(y_true_test)
y_pred_arr = np.array(y_pred_test)
n_test     = len(y_true_arr)

boot = {"Accuracy": [], "F1 (Macro)": [], "Precision (Macro)": [], "Recall (Macro)": [], "Specificity": []}
per_class_f1 = {name: [] for name in class_names}

def compute_specificity_boot(y_t, y_p, num_classes):
    cm = confusion_matrix(y_t, y_p, labels=list(range(num_classes)))
    specs = []
    for i in range(num_classes):
        TP = cm[i, i]; FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP; TN = cm.sum() - (TP + FP + FN)
        specs.append(TN / (TN + FP) if (TN + FP) > 0 else 0.0)
    return float(np.mean(specs))

for _ in range(N_BOOTSTRAP):
    idx = rng.integers(0, n_test, size=n_test)
    y_t, y_p = y_true_arr[idx], y_pred_arr[idx]
    boot["Accuracy"].append(accuracy_score(y_t, y_p))
    boot["F1 (Macro)"].append(f1_score(y_t, y_p, average="macro", zero_division=0))
    boot["Precision (Macro)"].append(precision_score(y_t, y_p, average="macro", zero_division=0))
    boot["Recall (Macro)"].append(recall_score(y_t, y_p, average="macro", zero_division=0))
    boot["Specificity"].append(compute_specificity_boot(y_t, y_p, NUM_CLASSES))
    pc = f1_score(y_t, y_p, average=None, zero_division=0, labels=list(range(NUM_CLASSES)))
    for ci, name in enumerate(class_names):
        per_class_f1[name].append(pc[ci])

def ci95(arr): return float(np.percentile(arr, 2.5)), float(np.percentile(arr, 97.5))

short_names = [n.split("(")[-1].replace(")", "").strip() if "(" in n else n
               for n in class_names]

W = 70
print("\n" + "=" * W)
print(f"  PAD-UFES-20 — C1 Tabular Only — Bootstrap CI (n={N_BOOTSTRAP})")
print("=" * W)
print(f"{'Metric':<22} {'Mean':>7}  {'Std':>7}  {'CI Lower':>10}  {'CI Upper':>10}")
print("=" * W)
for metric, values in boot.items():
    arr = np.array(values); lo, hi = ci95(arr)
    print(f"{metric:<22} {arr.mean():>7.4f}  {arr.std():>7.4f}  {lo:>10.4f}  {hi:>10.4f}")
print("-" * W)
print(f"\n{'Class':<12} {'Mean':>7}  {'Std':>7}  {'CI Lower':>10}  {'CI Upper':>10}")
print("-" * W)
for sname, name in zip(short_names, class_names):
    arr = np.array(per_class_f1[name]); lo, hi = ci95(arr)
    print(f"{sname:<12} {arr.mean():>7.4f}  {arr.std():>7.4f}  {lo:>10.4f}  {hi:>10.4f}")
print("=" * W)

fig, axes = plt.subplots(1, 5, figsize=(18, 6))
fig.subplots_adjust(top=0.82, wspace=0.4)
for ax, (label, values) in zip(axes, boot.items()):
    arr = np.array(values); lo, hi = ci95(arr); mn, sd = arr.mean(), arr.std()
    ax.boxplot(arr, widths=0.5, patch_artist=True,
               boxprops=dict(facecolor="#4c72b0", alpha=0.75),
               medianprops=dict(color="white", linewidth=2),
               whiskerprops=dict(linewidth=1.2), capprops=dict(linewidth=1.2),
               flierprops=dict(marker="o", markersize=3, markerfacecolor="gray", alpha=0.5))
    ax.set_title(label, fontsize=10, fontweight="bold", pad=8)
    ax.set_xticks([]); ax.set_ylabel("Score", fontsize=9)
    ax.grid(axis="y", alpha=0.3, linestyle="--")
    ax.text(1, ax.get_ylim()[1], f"μ={mn:.4f}\nσ={sd:.4f}\n95%CI [{lo:.4f},{hi:.4f}]",
            ha="center", va="top", fontsize=7.5, color="#222222",
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightyellow", edgecolor="gray", alpha=0.8))
plt.suptitle(f"PAD-UFES-20 C1 — Bootstrap Metric Distributions (n={N_BOOTSTRAP}) — Test Set",
             fontsize=13, fontweight="bold", y=0.97)
plt.savefig("/kaggle/working/pad_c1_bootstrap_boxplots.png", dpi=150, bbox_inches="tight")
plt.show()

means  = np.array([np.mean(per_class_f1[n]) for n in class_names])
stds   = np.array([np.std(per_class_f1[n])  for n in class_names])
lo_arr = np.array([ci95(per_class_f1[n])[0] for n in class_names])
hi_arr = np.array([ci95(per_class_f1[n])[1] for n in class_names])

fig, ax = plt.subplots(figsize=(13, 6))
x = np.arange(NUM_CLASSES)
bars = ax.bar(x, means, yerr=[means - lo_arr, hi_arr - means], capsize=6,
              color="#4c72b0", alpha=0.8, ecolor="#333333",
              error_kw={"linewidth": 1.8, "capthick": 1.8}, width=0.55)
ax.set_xticks(x)
ax.set_xticklabels(short_names, fontsize=11)
ax.set_ylabel("F1 Score", fontsize=11); ax.set_ylim(0, 1.15)
ax.set_title("PAD-UFES-20 C1 — Per-Class F1 Mean ± 95% CI (Bootstrap)",
             fontsize=13, fontweight="bold")
ax.grid(axis="y", alpha=0.3, linestyle="--"); ax.spines[["top", "right"]].set_visible(False)
for bar, m, s, hi in zip(bars, means, stds, hi_arr):
    ax.text(bar.get_x() + bar.get_width() / 2, hi + 0.025,
            f"{m:.3f}\n±{s:.3f}", ha="center", va="bottom", fontsize=9, color="#111111")
plt.tight_layout()
plt.savefig("/kaggle/working/pad_c1_bootstrap_perclass_f1.png", dpi=150, bbox_inches="tight")
plt.show()
```
